```python
import math

n = 500
media = 70
desviacion = 4

z1 = (68 - media) / desviacion
z2 = (72 - media) / desviacion

def normal(z):
    return (1 + math.erf(z / math.sqrt(2))) / 2

probabilidad = normal(z2) - normal(z1)

estudiantes = n * probabilidad

print("DISTRIBUCION NORMAL")
print("----------------------------")
print("Media:", media, "kg")
print("Desviacion estandar:", desviacion, "kg")
print("Probabilidad:", round(probabilidad, 4))
print("Probabilidad en porcentaje:", round(probabilidad * 100, 2), "%")
print("Cantidad de estudiantes:", round(estudiantes))
```
