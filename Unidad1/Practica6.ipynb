```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import math

pd.set_option('display.max_columns',None)
df=pd.read_csv('autos2.csv')
df.head()
```

```python
df['price']
pd.to_numeric(df['price'],
errors='coerce')

precios=df['price'].dropna()

minimo=precios.min()
maximo=precios.max()
moda=precios.mode()
media=precios.mean()
mediana=precios.median()
desviación=precios.std()

print("ESTADISTICAS DESCRIPTIVAS DEL PRECIO")
print("-------------------------------------")
print("Valor minimo: ",minimo)
print("Valor maximo: ",maximo)
print("Moda: ",moda.tolist())
print("Media: ",media)
print("Mediana: ",mediana)
print("Desviación estandar: ",desviación)
```

```python
print("ESTADÍSTICOS DESCRIPTIVOS")
print("--------------------------")
print(precios.describe())
```

```python
n=len(precios)
bins_sturges=math.ceil(1+math.log2(n))

print("Cantidad de datos:",n)
print("Cantidad optima de bins según Sturges: ",bins_sturges)
```

```python
plt.figure(figsize=(10,6))
plt.hist(precios,bins=bins_sturges,
         edgecolor='black')
plt.title('Histograma de Precios de Automóviles')
plt.xlabel('Precio del automóvil')
plt.ylabel('Freecuencia')
plt.grid(axis='y',alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x='body-style',
            y='price')
plt.title('Distribución del Precio según el tipo de vehiculo')
plt.xlabel('Tipo de vehiculo')
plt.ylabel('Precio del vehiculo')
plt.grid(axis='y',alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x='drive-wheels',
            y='price')
plt.title('Distribución del Precio según el tipo de tracción')
plt.xlabel('Tipo de tracción')
plt.ylabel('Precio')
plt.grid(axis='y',alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df,x='engine-location',
            y='price')
plt.title('Distribución del Precio según la ubicación del motor')
plt.xlabel('Ubicación del motor')
plt.ylabel('Precio')
plt.grid(axis='y',alpha=0.3)

plt.show()
```

El análisis permitio conocer la distribución y comportamiento de los precios de los automóviles mediante medidas estadísticas,histogramas y diagramas de caja.Se observaron diferencias en los precios según el tipo de vehiculo,tracción y ubicación del motor,además de algunos valores atípicos.
