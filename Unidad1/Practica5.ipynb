```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
covid=pd.read_csv("20-06-30-COVID19MEXICO - codificado.csv",encoding="ISO-8859-1")
```

```python
covid.info()
```

```python
conteo_edades=covid['EDAD'].value_counts().sort_index()
plt.bar(conteo_edades.index,conteo_edades.values,color='skyblue',
        edgecolor='black',alpha=0.6)
plt.title("Pruebas COVID por Edad")
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.show()

```

```python
media=covid.EDAD.mean()
rangos=[0,20,30,40,50,60,70,80,90,100,110,120]
plt.hist(covid.EDAD,bins=rangos,edgecolor='black')
plt.title("Hostograma de Pruebas de COVID por grupos de Edad")
plt.axvline(media,color='y',linestyle='dashed',linewidth=2,label=f'Media Edad={media:.2f}')
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.legend()
```

```python
conteo_sexo=covid['SEXO'].value_counts()
plt.bar(conteo_sexo.index.astype(str),conteo_sexo.values,
        color='lightblue',edgecolor='black',alpha=0.7)
plt.title("Pruebas COVID por sexo")
plt.xlabel("Sexo")
plt.ylabel("Frecuencia")
plt.show()
```

```python
conteo_tipo_paciente=covid['TIPO_PACIENTE'].value_counts()
plt.bar(conteo_tipo_paciente.index.astype(str),conteo_tipo_paciente.values,
        color='lightblue',edgecolor='black',alpha=0.7)
plt.title("Pruebas COVID por Tipo Paciente")
plt.xlabel("Tipo Paciente")
plt.ylabel("Frecuencia")
plt.show()
```

```python
sns.set_style("whitegrid")
sns.boxplot(x="SEXO",y="EDAD",data=covid,palette="pastel")
plt.title("Distribución de  edad por sexo",fontsize=14,fontweight="bold")
plt.xlabel("Sexo",fontsize=12)
plt.ylabel("Edad",fontsize=12)
plt.show()
```
