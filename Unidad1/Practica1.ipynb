```python
#Lista de Calificaciones del grupo A de probabilidad y estadistica
grupoA=[70,70,65,90,100,95,45,80,73,20,100,96]
print(grupoA)
```

```python
#Sacar la media de las calificaciones con una formula
promedio=sum(grupoA)/len(grupoA)
print(promedio)
```

```python
from statistics import mean
#Utilizar la función mean que esta en la libreria statisticspara sacr el promedio
promedio2=mean(grupoA)
print(promedio2)
```

```python

from statistics import median
#Determinar el valor de la mediana con la funcióon median
mediana=median(grupoA)
print(mediana)
```
