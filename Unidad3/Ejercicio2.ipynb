```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 10, 20, 50])
p = np.array([0.50, 0.30, 0.15, 0.05])

suma_p = np.sum(p)
f_x = np.cumsum(p)

e_x = np.sum(x * p)
var_x = np.sum(((x - e_x)**2) * p)
std_x = np.sqrt(var_x)

df = pd.DataFrame({'Premio (X)': x, 'P(X)': p, 'F(X)': f_x})

print(f"1. Suma de probabilidades: {suma_p:.2f}")
print("\n2. Distribucion de probabilidad y acumulada:")
print(df.to_string(index=False))
print(f"\n3. Valor Esperado E(X): {e_x:.2f}")
print(f"   Varianza: {var_x:.4f}")
print(f"   Desviacion Estandar: {std_x:.4f}")

plt.figure(figsize=(7, 4))
x_plot = np.concatenate(([x[0] - 5], x, [x[-1] + 5]))
y_plot = np.concatenate(([0], f_x, [1.0]))
plt.step(x_plot, y_plot, where='post', color='green')
plt.scatter(x, f_x, color='red')
plt.title('Ejercicio 2: Funcion Acumulada F(x)')
plt.xlabel('Monto en pesos (X)')
plt.ylabel('F(x)')
plt.ylim(-0.05, 1.1)
plt.xticks(x)
plt.grid(True)
plt.show()

print(f"• Significado del valor esperado: Representa la ganancia promedio por boleto a largo plazo (${e_x:.2f} pesos).")
print(f"• ¿Es favorable?: Solo si el costo del boleto es menor a ${e_x:.2f} pesos. El 50% de las veces no se gana nada.")
```
