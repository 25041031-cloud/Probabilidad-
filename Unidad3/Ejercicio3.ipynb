```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3, 4, 5])
p = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])

suma_p = np.sum(p)
f_x = np.cumsum(p)

e_x = np.sum(x * p)
var_x = np.sum(((x - e_x)**2) * p)
std_x = np.sqrt(var_x)

df = pd.DataFrame({'Unidades Vendidas (X)': x, 'P(X)': p, 'F(X)': f_x})

print(f"1. Suma de probabilidades: {suma_p:.2f}")
print("\n2. Distribucion de probabilidad y acumulada:")
print(df.to_string(index=False))
print(f"\n3. Valor Esperado E(X): {e_x:.2f}")
print(f"   Varianza: {var_x:.4f}")
print(f"   Desviacion Estandar: {std_x:.4f}")

plt.figure(figsize=(7, 4))
x_plot = np.concatenate(([x[0] - 1], x, [x[-1] + 1]))
y_plot = np.concatenate(([0], f_x, [1.0]))
plt.step(x_plot, y_plot, where='post', color='purple')
plt.scatter(x, f_x, color='red')
plt.title('Ejercicio 3: Funcion Acumulada F(x)')
plt.xlabel('Unidades vendidas (X)')
plt.ylabel('F(x)')
plt.ylim(-0.05, 1.1)
plt.xticks(x)
plt.grid(True)
plt.show()

print(f"• Ventas esperadas: Se espera vender {e_x:.2f} unidades por semana en promedio.")
print(f"• Estabilidad de la demanda: La demanda es estable (desviacion estandar de {std_x:.2f}), concentrandose la mayor parte de las semanas entre 2 y 4 unidades.")
```
